```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
loss_history = np.load("loss_history.npy")
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(loss_history) + 1),
    loss_history,
    label="Training Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.title("Training Loss over Epochs")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
