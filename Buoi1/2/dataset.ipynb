# Adzuna's Job Dataset


**1. Overview**  
Dataset thu thập thông tin các nghề thuộc lĩnh vực IT tại 2 khu vực New York(Mỹ) và London(Anh)
Dataset bao gồm 4375 mẫu với 18 thuộc tính.

**2. Data Source**  
Dữ liệu được thu thập từ trang web [Adzuna](https://www.adzuna.com/), một công cụ tìm kiếm công việc trực tuyến chuyên thu thập các tin tuyển dụng từ nhiều nguồn khác nhau.
Dữ liệu được thu thập thông qua API chính thức của [Adzuna API](https://developer.adzuna.com/)

**3. Data Dictionary**  
| STT | Tên thuộc tính | Kiểu dữ liệu | Ý nghĩa | Ví dụ |
|---:|---|---|---|---|
| 1 | id | int64 | Mã tin tuyển dụng | 5869886312 |
| 2 | title | str | Tên nghề nghiệp | Scientific Software Engineer |
| 3 | company_name | str | Tên công ty | Pertemps Reading |
| 4 | location_name | str | Khu vực tuyển dụng | "South East London, London" |
| 5 | latitude | float64 | Vĩ độ | 51.451818 |
| 6 | longitude | float64 | Kinh độ | -0.02806 |
| 7 | search_country | str  | Quốc gia tìm kiếm | gb |
| 8 | search_location | str | Vị trí tìm kiếm | London |
| 9 | search_keyword | str | Tù khóa tìm kiếm |
| 10 | category_name | str | Ngành nghề | IT Jobs |
| 11 | salary_min | float64 | Mức lương tối thiểu | 95000.0 |
| 12 | salary_max | float64 | Mức lương tối đa | 95000.0 |
| 13 | salary_is_predicted | int64 | Mức lương có được dự đoán | 0 |
| 14 | contract_type | str | Loại hợp đồng lao động | permanent |
| 15 | contract_time | str | Thời gian lao động | full_time |
| 16 | created | str | Thời gian đăng tin | 2026-09-04T03:33:06Z |
| 17 | description | str | Mô tả công việc | "Scientific Software Engineer Salary…" |
| 18 | redirect_url | str | Link dẫn đến trang tin dụng của Adzuna | https://www.adzuna.co.uk/...|
